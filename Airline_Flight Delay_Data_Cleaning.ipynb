## 1. Import required libraries
Pandas is used for tabular data handling, NumPy for numerical operations, and Matplotlib is imported only if a basic visual check is needed.

```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

## 2. Upload and load the dataset
In Google Colab, upload the `14_flight_delays.csv` file when prompted. When this notebook is run outside Colab, it looks for the CSV in the same working folder.

## 3. Inspect the original dataset
Check column names, data types, missing values, duplicate rows and basic numeric summaries before making changes.

```python
print("Dataset shape:", df.shape)
print("\\nColumn names:")
print(df.columns.tolist())

print("\\nDataset information:")
df.info()

print("\\nMissing values before cleaning:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\\nExact duplicate rows before cleaning:", df.duplicated().sum())

print("\\nNumeric summary:")
display(df.describe(include="number"))
```

## 4. Clean text fields and standardize data types
Leading/trailing spaces are removed from text columns. Dates and numeric delay columns are converted to suitable data types. Category values are standardized without changing the meaning of the records.

```python
clean_df = df.copy()

# Remove leading/trailing spaces from text columns
text_cols = clean_df.select_dtypes(include=["object"]).columns
for col in text_cols:
    clean_df[col] = clean_df[col].astype("string").str.strip()

# Convert blank text values to missing values
clean_df = clean_df.replace(r"^\\s*$", pd.NA, regex=True)

# Convert date and numeric columns
clean_df["Flight_Date"] = pd.to_datetime(clean_df["Flight_Date"], errors="coerce")
numeric_cols = ["Departure_Delay_Mins", "Arrival_Delay_Mins"]
for col in numeric_cols:
    clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")

# Standardize yes/no fields
for col in ["Canceled", "Diverted"]:
    clean_df[col] = clean_df[col].astype("string").str.strip().str.title()

# Standardize text category formatting
for col in ["Airline", "Origin", "Destination", "Departure_Time", "Delay_Cause"]:
    clean_df[col] = clean_df[col].astype("string").str.strip()

print("Text and data-type standardization completed.")
print(clean_df.dtypes)
```

## 5. Remove duplicate records
Exact duplicate rows are removed. The original dataset contains repeated complete records, so this step removes only exact duplicates and preserves distinct flights.

```python
rows_before = len(clean_df)
clean_df = clean_df.drop_duplicates().copy()
duplicates_removed = rows_before - len(clean_df)

print("Rows before duplicate removal:", rows_before)
print("Exact duplicate rows removed:", duplicates_removed)
print("Rows after duplicate removal:", len(clean_df))
print("Duplicate Flight_ID values remaining:", clean_df["Flight_ID"].duplicated().sum())
```

## 6. Handle missing values
- `Delay_Cause`: missing reasons are labelled **Not Recorded**. The cause is not guessed.
- `Arrival_Delay_Mins`: missing numeric values for operated flights are filled using the median arrival delay for the same airline. If an airline has no available median, the overall median is used.
- A new flag, `Arrival_Delay_Was_Missing`, records which arrival-delay values were imputed.
- The departure-delay field is not imputed unless missing values are found; if any exist, the same airline-median method is applied and a flag is added.

```python
# Record which arrival-delay values were missing before imputation
clean_df["Arrival_Delay_Was_Missing"] = clean_df["Arrival_Delay_Mins"].isna()

# Fill missing arrival delay with the median for that airline
arrival_medians = clean_df.groupby("Airline")["Arrival_Delay_Mins"].transform("median")
overall_arrival_median = clean_df["Arrival_Delay_Mins"].median()
clean_df["Arrival_Delay_Mins"] = clean_df["Arrival_Delay_Mins"].fillna(arrival_medians)
clean_df["Arrival_Delay_Mins"] = clean_df["Arrival_Delay_Mins"].fillna(overall_arrival_median)

# If departure-delay values are missing, preserve a flag and impute by airline median
if clean_df["Departure_Delay_Mins"].isna().any():
    clean_df["Departure_Delay_Was_Missing"] = clean_df["Departure_Delay_Mins"].isna()
    dep_medians = clean_df.groupby("Airline")["Departure_Delay_Mins"].transform("median")
    clean_df["Departure_Delay_Mins"] = clean_df["Departure_Delay_Mins"].fillna(dep_medians)
    clean_df["Departure_Delay_Mins"] = clean_df["Departure_Delay_Mins"].fillna(
        clean_df["Departure_Delay_Mins"].median()
    )

# Do not guess missing causes
clean_df["Delay_Cause"] = clean_df["Delay_Cause"].fillna("Not Recorded")

print("Missing arrival-delay values imputed:", int(clean_df["Arrival_Delay_Was_Missing"].sum()))
print("Missing delay causes labelled Not Recorded:", int((clean_df["Delay_Cause"] == "Not Recorded").sum()))
print("\\nMissing values after handling:")
display(clean_df.isnull().sum().to_frame("Remaining Missing Values"))
```

## 7. Final validation
Verify the cleaned dataset has no exact duplicate rows, no duplicate flight IDs, consistent types and the expected number of records.

```python
print("Final dataset shape:", clean_df.shape)
print("Exact duplicate rows remaining:", clean_df.duplicated().sum())
print("Duplicate Flight_ID values remaining:", clean_df["Flight_ID"].duplicated().sum())
print("\\nFinal data types:")
print(clean_df.dtypes)
print("\\nFinal missing-value count:")
display(clean_df.isnull().sum().to_frame("Missing Values"))
print("\\nPreview of cleaned dataset:")
display(clean_df.head())
```

The cleaned file is saved as `cleaned_flight_delays.csv`. In Google Colab, the file is automatically downloaded.

```python
output_file = "cleaned_flight_delays.csv"
clean_df.to_csv(output_file, index=False)

print("Cleaned dataset saved as:", output_file)

try:
    from google.colab import files
    files.download(output_file)
except ImportError:
    print("Outside Google Colab: file saved in the current working directory.")
```
